## Ejercicio 2
*(Exercise 2)*
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

*(Write a program to convert from degrees to radians. You must use `input`. Remember the conversion formula is:)*

radianes = grados*(pi/180)

````python
import math
# Pedir los grados al usuario / Ask the user for degrees
grados = float(input("Enter degrees: "))
# Aplicar la fórmula de conversión / Apply the conversion formula
radianes = grados * (math.pi / 180)
print(radianes)
````

## Ejercicio 6
*(Exercise 6)*
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

*(Same program as in 5, but this time with three inputs and two outputs. One output tells us if all three are equal, and the other if at least two inputs are equal)*

````python
# Recoger tres valores del usuario / Collect three values from the user
i1 = input("Enter first value: ")
i2 = input("Enter second value: ")
i3 = input("Enter third value: ")

# Comprobar si todos son iguales / Check if all three are equal
todos_iguales = (i1 == i2 == i3)
# Comprobar si al menos dos son iguales / Check if at least two are equal
al_menos_dos = (i1 == i2) or (i1 == i3) or (i2 == i3)

print("Todos son iguales:", todos_iguales)
print("Al menos dos son iguales:", al_menos_dos)
````

*(Create a program that collects two inputs. It must check whether their sum is equal to, greater than, or less than 10)*

## Ejercicio 10
*(Exercise 10)*
Para el siguiente string se pide imprimir por pantalla los siguientes casos
*(For the following string, print to the screen the following cases)*

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas *(Convert everything to uppercase)*
2. Pásalo todo a minúsculas *(Convert everything to lowercase)*
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula *(Only the first letter of each word in uppercase, the rest in lowercase)*
4. Crea una lista dividiéndolo por sus espacios *(Create a list by splitting it by spaces)*
5. Sustituye las comas `,` por puntos y comas `;` *(Replace commas `,` with semicolons `;`)*
6. Elimina las `a` minúsculas *(Remove lowercase `a` letters)*

````python
texto = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

# 1. Todo en mayúsculas / All uppercase
print(texto.upper())
# 2. Todo en minúsculas / All lowercase
print(texto.lower())
# 3. Primera letra de cada palabra en mayúscula / First letter of each word capitalized
print(texto.title())
# 4. Dividir por espacios en una lista / Split by spaces into a list
print(texto.split())
# 5. Reemplazar comas por punto y coma / Replace commas with semicolons
print(texto.replace(",", ";"))
# 6. Eliminar todas las 'a' minúsculas / Remove all lowercase 'a'
print(texto.replace("a", ""))
````

## Ejercicio 11
*(Exercise 11)*
1. Crea una lista con 3 elementos numéricos *(Create a list with 3 numeric elements)*
2. Añade un cuarto elemento *(Add a fourth element)*
3. Calcula la suma de todos *(Calculate the sum of all)*
4. Elimina el segundo elemento de la lista *(Remove the second element from the list)*
5. Añade otro elemento en la posicion 3 de la lista *(Add another element at position 3 of the list)*
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías. *(Create another list with 4 elements and concatenate it to the existing one.)*

````python
# 1. Crear lista con 3 elementos numéricos / Create list with 3 numeric elements
lista = [2, 3, 4]
# 2. Añadir un cuarto elemento / Add a fourth element
lista.append(1)
# 3. Suma de todos los elementos / Sum of all elements
print(sum(lista))
# 4. Eliminar el segundo elemento (índice 1) / Remove the second element (index 1)
lista.pop(1)
print(lista)
# 5. Insertar en la posición 3 (índice 2) / Insert at position 3 (index 2)
lista.insert(2, 5)
print(lista)
# 6. Crear otra lista y concatenar / Create another list and concatenate
lista2 = [6, 7, 8, 9]
lista = lista + lista2
print(lista)
````
