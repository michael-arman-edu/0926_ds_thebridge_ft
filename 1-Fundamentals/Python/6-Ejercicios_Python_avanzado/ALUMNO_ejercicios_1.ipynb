```python
lista_carrusel = ["barco", "bomberos", "avioneta"]

def coloca(cochecito):
    if cochecito == "barco":
        return lista_carrusel
    elif cochecito == "bomberos" or cochecito == "avioneta":
        elemento = lista_carrusel.pop(0)
        lista_carrusel.append(elemento)
    if cochecito == "avioneta":
        elemento = lista_carrusel.pop(0)
        lista_carrusel.append(elemento)
    return lista_carrusel

# Define the carrusel function with parameters: cochecito (string) and vueltas (int)
# Use a while loop that counts from 1 to vueltas
# On each iteration call coloca(cochecito) to reorder the list
# Print the current round number and the list joined with " - "
# Increment the counter each iteration

#carrusel("avioneta", 3)
```
