## Ejercicio 1
*(Exercise 1)*
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana.
*(Write a function that converts numbers from 1 to 7 into the names of the days of the week.)*

```python
# Diccionario que mapea número a nombre del día / Dictionary mapping number to day name
def dia_semana(numero):
    dias = {
        1: "Lunes",
        2: "Martes",
        3: "Miércoles",
        4: "Jueves",
        5: "Viernes",
        6: "Sábado",
        7: "Domingo"
    }
    # Devolver el nombre si existe, o None si no / Return the name if it exists, or None if not
    return dias.get(numero, None)

print(dia_semana(1))
print(dia_semana(7))
```

```python
# Probar con un número fuera de rango / Test with an out-of-range number
resultado = dia_semana(10)
if resultado is None:
    print("El día no existe")
else:
    print(resultado)
```

## Ejercicio 2
*(Exercise 2)*
Crear una función que replique el comportamiento de la pirámide invertida.
*(Create a function that replicates the inverted pyramid behavior.)*

```python
# Función que imprime una pirámide invertida de n filas / Function that prints an inverted pyramid of n rows
def piramide(n):
    for i in range(n, 0, -1):
        fila = ""
        # Construir cada fila con números del 1 hasta i / Build each row with numbers from 1 to i
        for j in range(1, i + 1):
            fila += " " + str(j)
        print(fila)

# Probar con 4 y 6 filas / Test with 4 and 6 rows
piramide(4)
piramide(6)
```

```python
# Llamar a la función sin argumento para comprobar que devuelve None / Call without argument to check it returns None
print(piramide(0))
```

## Ejercicio 3
*(Exercise 3)*
Función que compara dos números con tres posibles salidas.
*(Function that compares two numbers with three possible outputs.)*

```python
# Comparar dos números e imprimir el resultado / Compare two numbers and print the result
def comparar(a, b):
    if a > b:
        print(f"{a} es mayor que {b}")
    elif b > a:
        print(f"{b} es mayor que {a}")
    else:
        # Caso de igualdad / Equality case
        print("Son iguales")

comparar(4, 3)
comparar(5, 5)
```

## Ejercicio 4
*(Exercise 4)*
Función contador de letras (mayúsculas y minúsculas).
*(Letter counter function, both uppercase and lowercase.)*

```python
# Contar apariciones de una letra ignorando mayúsculas / Count occurrences of a letter ignoring case
def contar_letra(texto, letra):
    # Convertir todo a minúsculas para comparar sin distinción / Convert to lowercase for case-insensitive comparison
    return texto.lower().count(letra.lower())

print(contar_letra("Hola Mundo", "o"))
```

## Ejercicio 5
*(Exercise 5)*
Función que devuelve un diccionario con el conteo de todas las letras de un string.
*(Function that returns a dictionary with the count of all letters in a string.)*

```python
# Contar cada carácter del string y almacenarlo en un diccionario / Count each character and store in a dictionary
def contar_letras(texto):
    conteo = {}
    for char in texto:
        # Incrementar el contador o inicializarlo a 1 / Increment the counter or initialize it to 1
        conteo[char] = conteo.get(char, 0) + 1
    return conteo

print(contar_letras("en esta clase practicamos python y somos muy chulos"))
```

```python
# Probar con otra cadena / Test with another string
print(contar_letras("hola hola"))
```

## Ejercicio 6
*(Exercise 6)*
Función que añade o elimina elementos de una lista.
*(Function that adds or removes elements from a list.)*

```python
# Función con argumento por defecto para el elemento / Function with default argument for the element
def modificar_lista(lista, comando, elemento=None):
    if comando == "add":
        if elemento is None:
            # Si no se da elemento, añadir 9999 como aviso / If no element given, add 9999 as a warning
            lista.append(9999)
            print("Añadido el 9999 porque no se introdujo valor en el argumento elemento")
        else:
            lista.append(elemento)
    elif comando == "remove":
        if elemento is None:
            print("No ha introducido el elemento")
        elif elemento not in lista:
            # Avisar si el elemento no existe en la lista / Warn if the element does not exist in the list
            print(f"El elemento {elemento} no existe en la lista")
        else:
            lista.remove(elemento)
    else:
        print("error")
    return lista

mi_lista = [1, 2, 3]
print(modificar_lista(mi_lista, "add", 9))
print(modificar_lista(mi_lista, "add"))
print(modificar_lista(mi_lista, "remove", 2))
print(modificar_lista(mi_lista, "remove"))
print(modificar_lista([1, 2, 3], "remove", 10))
print(modificar_lista([1, 2, 3], "xyz"))
```

Función que recibe un número arbitrario de palabras y devuelve una frase.
*(Function that receives an arbitrary number of words and returns a sentence.)*

```python
# *args permite pasar un número variable de argumentos / *args allows passing a variable number of arguments
def formar_frase(*palabras):
    # Unir todas las palabras con un espacio / Join all words with a space
    return " ".join(palabras)

print(repr(formar_frase("Hola", "me", "llamo", "Alberto")))
```

## Ejercicio 8
*(Exercise 8)*
Función recursiva para obtener el enésimo número de Fibonacci.
*(Recursive function to get the nth Fibonacci number.)*

```python
# Casos base: fibonacci(1)=1 y fibonacci(2)=1 / Base cases: fibonacci(1)=1 and fibonacci(2)=1
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1 or n == 2:
        return 1
    # Llamada recursiva: suma de los dos anteriores / Recursive call: sum of the two previous
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(1))
print(fibonacci(2))
print(fibonacci(4))
print(fibonacci(5))
print(fibonacci(8))
```

## Ejercicio 9
*(Exercise 9)*
Funciones para calcular áreas de cuadrado, triángulo y círculo.
*(Functions to calculate areas of a square, triangle and circle.)*

```python
import math

# Área del cuadrado: lado al cuadrado / Square area: side squared
def area_cuadrado(lado):
    return lado ** 2

# Área del triángulo: base por altura entre dos / Triangle area: base times height divided by two
def area_triangulo(base, altura):
    return (base * altura) / 2

# Área del círculo: pi por radio al cuadrado / Circle area: pi times radius squared
def area_circulo(radio):
    return math.pi * radio ** 2
```

```python
# Dos círculos de radio 10 + un triángulo de base 3 y altura 7 / Two circles of radius 10 + triangle base 3 height 7
resultado1 = 2 * area_circulo(10) + area_triangulo(3, 7)
print(resultado1)

# Cuadrado lado 10 + 3 círculos + 5 triángulos / Square side 10 + 3 circles + 5 triangles
resultado2 = area_cuadrado(10) + area_circulo(4) + 2 * area_circulo(6) + 5 * area_triangulo(2, 4)
print(resultado2)
```
