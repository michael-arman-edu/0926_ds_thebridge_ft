## Ejercicio 1
*(Exercise 1)*
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana.
*(Write a function that converts numbers from 1 to 7 into the names of the days of the week.)*

```python
# Define a function that takes a number as input
# Create a dictionary mapping numbers 1-7 to day names
# Check if the number is in the dictionary
# If yes, return the day name; if no, return an error message

def day_of_week(num):
    days = {
        1: "Monday",
        2: "Tuesday",
        3: "Wednesday",
        4: "Thursday",
        5: "Friday",
        6: "Saturday",
        7: "Sunday"
    }
    return days[num] if num in days else "Please enter a number between 1 and 7."

# test
print(day_of_week(1))
print(day_of_week(5))
print(day_of_week(9))
```

## Ejercicio 2
*(Exercise 2)*
Crear una función que replique el comportamiento de la pirámide invertida.
*(Create a function that replicates the inverted pyramid behavior.)*

```python
# Define a function that takes n (number of rows) as input
# Loop from n down to 1 so each row has one less star
# On each iteration, print i stars joined together
# The first row has n stars, last row has 1 star

def inverted_pyramid(n):
    for i in range(n, 0, -1):
        print("*" * i)

# test
inverted_pyramid(5)
```

## Ejercicio 3
*(Exercise 3)*
Función que compara dos números con tres posibles salidas.
*(Function that compares two numbers with three possible outputs.)*

```python
# Function that compares two numbers with three possible outputs

def compare_numbers(num1, num2):
    if num1 > num2:
        return f"{num1} is greater than {num2}"
    elif num1 < num2:
        return f"{num1} is less than {num2}"
    else:
        return f"{num1} is equal to {num2}"

# test the function with some examples

print(compare_numbers(5, 3)) 
print(compare_numbers(2, 4))
print(compare_numbers(7, 7))




```

## Ejercicio 4
*(Exercise 4)*
Función contador de letras (mayúsculas y minúsculas).
*(Letter counter function, both uppercase and lowercase.)*

```python
# Letter counter function, both uppercase and lowercase

def letter_counter(string):
    uppercase_count = 0
    lowercase_count = 0
    for char in string:
        if char.isupper():
            uppercase_count += 1
        elif char.islower():
            lowercase_count += 1
    return ('uppercase count:', uppercase_count, 'lowercase count:', lowercase_count)

# test the function with examples
letters = "Hola Buenos Dias"
letter_counter(letters) 
```

## Ejercicio 5
*(Exercise 5)*
Función que devuelve un diccionario con el conteo de todas las letras de un string.
*(Function that returns a dictionary with the count of all letters in a string.)*

```python
# Define a function that takes a string as input
# Create an empty dictionary to store letter counts
# Loop through each character in the string
# Skip non-letter characters (spaces, punctuation)
# If letter already in dict, increment count by 1
# If letter not in dict, add it with count of 1
# Return the dictionary

def letter_count(string):
    counts = {}
    for char in string:
        if char.isalpha():
            if char in counts:
                counts[char] += 1
            else:
                counts[char] = 1
    return counts

# test
print(letter_count("Hola Buenos Dias"))
```

## Ejercicio 6
*(Exercise 6)*
Función que añade o elimina elementos de una lista.
*(Function that adds or removes elements from a list.)*

```python
# Define a function that takes a list and an action ("add" or "remove")
# Also take the element to add or remove as a parameter
# If action is "add", append the element to the list
# If action is "remove", check if element exists then remove it
# Return the updated list

def modify_list(lst, action, element):
    if action == "add":
        lst.append(element)
    elif action == "remove":
        if element in lst:
            lst.remove(element)
        else:
            print(f"{element} not found in list")
    return lst

# test
my_list = [1, 2, 3]
print(modify_list(my_list, "add", 4))
print(modify_list(my_list, "remove", 2))
print(modify_list(my_list, "remove", 99))
```

## Ejercicio 7
*(Exercise 7)*
Función que recibe un número arbitrario de palabras y devuelve una frase.
*(Function that receives an arbitrary number of words and returns a sentence.)*

```python
# Use *args to accept any number of word arguments
# Join all the words with a space between them
# Return the resulting sentence as a string

def make_sentence(*args):
    return " ".join(args)

# test
print(make_sentence("Hello", "world"))
print(make_sentence("Python", "is", "great", "for", "data"))
```

## Ejercicio 8
*(Exercise 8)*
Función recursiva para obtener el enésimo número de Fibonacci.
*(Recursive function to get the nth Fibonacci number.)*

```python
# Define a recursive function that takes n as input
# Base case: if n is 0 return 0, if n is 1 return 1
# Recursive case: return fibonacci(n-1) + fibonacci(n-2)
# Each call breaks the problem down until it hits the base case

def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# test
print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(7))
print(fibonacci(10))
```

## Ejercicio 9
*(Exercise 9)*
Funciones para calcular áreas de cuadrado, triángulo y círculo.
*(Functions to calculate areas of a square, triangle and circle.)*

```python
# Import pi from math module for the circle area formula
# Define square_area: take side, return side squared
# Define triangle_area: take base and height, return base*height/2
# Define circle_area: take radius, return pi * radius squared

import math

def square_area(side):
    return side ** 2

def triangle_area(base, height):
    return base * height / 2

def circle_area(radius):
    return math.pi * radius ** 2

# test
print("Square area (side=4):", square_area(4))
print("Triangle area (base=6, height=3):", triangle_area(6, 3))
print("Circle area (radius=5):", round(circle_area(5), 2))
```
