## 2. Argumentos posicionales
*(2. Positional arguments)*

Ya sabes cómo crear funciones con un solo argumento. Tendrás la opción de implementarlas con todos los argumentos que quieras. Ahora bien, ten en cuenta dos cosas:
*(You already know how to create functions with a single argument. You can implement them with as many arguments as you want. However, keep two things in mind:)*

1. **El orden** de los argumentos. Cuando llamemos a la función, tenemos que seguir el mismo orden de argumentos que en la declaración de la función.
*(1. **The order** of arguments. When calling the function, we must follow the same order as in the function declaration.)*
2. **Son obligatorios**. Si los declaramos en la función, después al llamarla, tenemos que poner todos sus argumentos.
*(2. **They are mandatory**. If we declare them in the function, we must pass all arguments when calling it.)*

````python
def multiplica(x1, x2, x3, x4):
    return (x1 * x2 * x3) / x4

multiplica(4,6,7,2)
````

Fijate que los argumentos siguen un determinado orden: x1, x2, x3, x4. Cuando llamamos a la función, introduciremos 4 argumentos y la función los recogerá en ese orden. Asignará 4 a x1, 6 a x2, etc. Podemos también especificar el nombre del argumento en la llamada, lo que nos permite tener mayor flexibilidad en el orden.

````python
multiplica(x2 = 6, # Aqui pongo x2
            x1 = 4, # Y aqui x1
            x3 = 7,
            x4 = 2)
````

*(Keep in mind that these arguments are placed **at the end**.)*

## 5. Recursividad
*(5. Recursion)*

Una función se puede llamar a si misma en la propia declaración, como si de un bucle se tratase. Es un concepto algo complejo, pero elegante a la hora de implementar nuestros programas. La única parte negativa es que cuesta un poco comprender qué es lo que hace la función. Como todo, tiene sus ventajas y sus inconvenientes.
*(A function can call itself within its own declaration, as if it were a loop. It is a somewhat complex concept, but elegant when implementing our programs. The only downside is that it can be a bit hard to understand what the function is doing. Like everything, it has its advantages and disadvantages.)*

Calculemos el factorial de un numero *!n*
*(Let's calculate the factorial of a number *!n*)*

0! = 1
1! = 1 x 0! = 1 x 1 = 1
2! = 2 x 1! = 2 x 1 x 0! = 2 x 1 x 1 = 2
3! = 3! 3 x 2!


````python
# Lo podriamos calcular con un bucle
num_factorial = 3

output = 1
for i in range(num_factorial):
    output = output * (i + 1)
    print(i + 1)
print(output)
````

## 6. Documentar funciones
*(6. Documenting functions)*

Como ya vimos en el primer Notebook, hay que documentar el código en la medida de lo posible. En particular, es necesario documentar bien las funciones. porque muchas veces las importamos de otro sitio, las usamos porque funcionan, pero no sabemos muy bien que hacen. Es por ello, que en Python existe un atributo dentro de las funciones, módulos, métodos o clases, que permite acceder a "sus comentarios", a su documentación, donde nos indica qué es lo que hace.
*(As we saw in the first Notebook, we should document the code as much as possible. In particular, it is important to document functions well, because we often import them from somewhere else and use them without fully understanding what they do. That is why Python provides a special attribute inside functions, modules, methods or classes that allows access to their "comments" — their documentation — which tells us what they do.)*

Este atributo especial se llama *docstring*, y se accede mediante `nombre_funcion.__doc__`
*(This special attribute is called a *docstring*, and it is accessed via `function_name.__doc__`.)*

````python
# O mediante una funcion recursiva
def factorial(x):
    if x == 1:
        return 1
    else:
        return (x * factorial(x-1))


# 1! = 1
# 2! = 2 * 1 = 2 * 1!
# 3! = 3 * 2 * 1 = 3 * 2!
# 4! = 4 * 3 * 2 * 1 = 4 * 3!

# 200! = 200 * 199! 
# 199! = 199 * 198!
#...
# 1!


num_factorial = 3
print("El factorial de", num_factorial, "es", factorial(num_factorial ))
````

````python
def multiplica(x,y):
    """
    Funcion que multiplca los dos argumentos: x*y
    Inputs:
        x: float
        y: float
    
    Output:
        x * y: float
    """
    print("Empieza la funcion")
    # Mas comentarios
    
    return x*y

print(multiplica(2,2))
print(multiplica.__doc__)
````

Los comentarios que se ponen pueden ser de línea o multilínea. Para funciones sencillas puede ser suficiente con una sola línea de comentario, pero si fuesen más complejas, el *docstring* debería llevar la siguiente información:
*(The comments can be single-line or multi-line. For simple functions a single line of comment may be enough, but for more complex ones, the *docstring* should include the following information:)*
* Descripción de la función *(Description of the function)*
* Argumentos de entrada: nombre, tipos y qué es lo que hacen *(Input arguments: name, types and what they do)*
* Argumentos de salida: nombre, tipos y qué son *(Output arguments: name, types and what they are)*
