## Ejercicio 1
*(Exercise 1)*
Modelizar una tienda de electrodomésticos como clase con atributos, métodos y proyección de ventas.
*(Model an appliance store as a class with attributes, methods and sales projection.)*

```python
class Tienda:
    """Clase que representa una tienda de electrodomésticos. / Class representing an appliance store."""

    # Atributos de clase compartidos por todas las instancias / Class attributes shared by all instances
    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nombre, direccion, num_empleados, ventas_3_meses):
        # Atributos de instancia únicos para cada tienda / Instance attributes unique to each store
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_3_meses = ventas_3_meses

    def total_ventas(self):
        """Suma todas las ventas de los últimos 3 meses. / Sum all sales of the last 3 months."""
        return sum(self.ventas_3_meses)

    def media_ventas_empleado(self):
        """Media de ventas por empleado. / Average sales per employee."""
        # Dividir el total entre el número de empleados / Divide total by number of employees
        return self.total_ventas() / self.num_empleados

    def info(self):
        """Devuelve nombre y dirección como string. / Returns name and address as a string."""
        return f"Nombre tienda: {self.nombre}\nDireccion: {self.direccion}"

    def ventas_ultimo_mes(self):
        """Devuelve las ventas del último mes. / Returns last month's sales."""
        return self.ventas_3_meses[-1]

    def proyeccion_marketing(self, x):
        """Proyecta ventas según inversión en marketing. / Projects sales based on marketing investment."""
        if x < 1000:
            # Incremento del 20% para inversión menor de 1000 / 20% increase for investment below 1000
            factor = 1.2
        else:
            # Incremento del 50% para inversión mayor o igual a 1000 / 50% increase for investment >= 1000
            factor = 1.5
        self.ventas_3_meses = [v * factor for v in self.ventas_3_meses]
        return self.ventas_3_meses
```

```python
# Crear tres tiendas con datos inventados / Create three stores with made-up data
tienda1 = Tienda("mi tienda 1", "Avenida del electrodomestico", 3, [1000, 2000, 1500])
tienda2 = Tienda("mi tienda 2", "Calle Goya 10", 5, [2000, 3000, 2500])
tienda3 = Tienda("mi tienda 3", "Avenida de la Paz 5", 2, [500, 800, 600])

# Comprobar atributos y métodos de tienda1 / Check attributes and methods of tienda1
print(tienda1.tipo)
print(tienda1.abierta)
print(tienda1.nombre)
print(tienda1.direccion)
print(tienda1.num_empleados)
print()
print(tienda1.total_ventas())
print(tienda1.media_ventas_empleado())
print(tienda1.info())
print(tienda1.ventas_ultimo_mes())
print(tienda1.proyeccion_marketing(1000))
```

```python
# Ventas del último mes de todas las tiendas con for / Last month's sales of all stores with for
tiendas = [tienda1, tienda2, tienda3]
total_ultimo_mes = 0
for t in tiendas:
    total_ultimo_mes += t.ventas_ultimo_mes()
print(total_ultimo_mes)

# Imprimir nombres cuya dirección contiene 'Avenida' / Print names whose address contains 'Avenida'
for t in tiendas:
    if "Avenida" in t.direccion:
        print(t.nombre)
```

## Ejercicio 2
*(Exercise 2)*
Implementar una clase Perro con atributos fijos, métodos de movimiento y ladrido.
*(Implement a Dog class with fixed attributes, movement and bark methods.)*

```python
class Perro:
    """Clase que representa un perro. / Class representing a dog."""

    # Atributos de clase comunes a todos los perros / Class attributes common to all dogs
    patas = 4
    orejas = 2
    ojos = 2

    def __init__(self, raza, pelo="Marrón", dueño=None):
        # Atributos de instancia con valores por defecto / Instance attributes with default values
        self.raza = raza
        self.pelo = pelo
        self.dueño = dueño
        # El perro empieza parado / The dog starts stationary
        self.velocidad = 0

    def andar(self, aumento_velocidad):
        """Aumenta la velocidad del perro. / Increases the dog's speed."""
        self.velocidad += aumento_velocidad

    def parar(self):
        """Detiene al perro poniendo la velocidad a 0. / Stops the dog by setting speed to 0."""
        self.velocidad = 0

    def ladrar(self, mensaje):
        """Devuelve un ladrido con mensaje. / Returns a bark with a message."""
        return "GUAU!, " + mensaje
```

```python
# Crear un perro sin dueño / Create a dog without an owner
mi_perro = Perro("San Bernardo")

# Comprobar atributos / Check attributes
print(mi_perro.raza)
print(mi_perro.pelo)
print(mi_perro.dueño)
print(mi_perro.velocidad)
print(mi_perro.orejas)
print(mi_perro.patas)

# Hacer que el perro ande y comprobar velocidad / Make the dog walk and check speed
mi_perro.andar(5)
print(mi_perro.velocidad)

# Parar al perro / Stop the dog
mi_perro.parar()
print(mi_perro.velocidad)

# Hacer que el perro ladre / Make the dog bark
print()
print(repr(mi_perro.ladrar("quiero comer")))
```
