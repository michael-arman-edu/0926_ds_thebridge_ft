## Ejercicio 1
*(Exercise 1)*
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:

*(We are a company that has several appliance stores. We need a program to manage the stores, view their sales, how many employees there are, etc. We will model the store as a class with the following characteristics:)*

* Nombre clase: "Tienda" *(Class name: "Tienda")*
* Atributos comunes: *(Common attributes:)*
    * Tipo: "Electrodomésticos" *(Type: "Electrodomésticos")*
    * Abierta: True *(Open: True)*
* Atributos propios de cada tienda: *(Instance attributes:)*
    * Nombre - String *(Name - String)*
    * Dirección - String *(Address - String)*
    * Número de empleados - int *(Number of employees - int)*
    * Ventas ultimos 3 meses - Lista de 3 numeros *(Sales last 3 months - List of 3 numbers)*
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas. *(Method to calculate total sales for all months, returning a number.)*
* Método que calcula la media de ventas de los ultimos meses, por empleado *(Method that calculates the average sales for the last months, per employee)*
* Método que devuelve en un string el nombre de la tienda, junto con su dirección. *(Method that returns a string with the store name and its address.)*
* Método que obtiene las ventas del último mes. *(Method that gets the sales of the last month.)*
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento. *(Method for a sales projection if X money had been invested in marketing. If X < 1000, sales would have been (1.2 × sales); if X >= 1000, sales would have been (1.5 × sales). The method overwrites the sales attribute and returns the increased sales.)*

Se pide: *(Required:)*
1. Implementar la clase "Tienda" *(Implement the "Tienda" class)*
2. Crear tres tiendas con datos inventados *(Create three stores with made-up data)*
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...) *(Check in at least one of them everything implemented in the store class (attributes, average sales, sales/employee...))*
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for` *(Calculate the last month's sales for all stores using a `for` loop)*
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida" *(Print the names of stores whose address contains the string "Avenida")*
5. Seguro que ya has hecho este apartado... Documenta la clase :) *(I'm sure you've already done this part… Document the class :))*

## Ejercicio 2
*(Exercise 2)*
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
*(In this exercise we will implement a *Dog* class in Python. The class has the following characteristics:)*
* Cosas que sabemos seguro que tiene un perro *(Things we know for sure a dog has)*
    * Tiene 4 patas *(Has 4 legs)*
    * 2 orejas *(2 ears)*
    * 2 ojos *(2 eyes)*
    * Una velocidad de 0. Por defecto, el perro está parado *(A speed of 0. By default, the dog is stationary)*
* Cuando se inicialice: *(When initialized:)*
    * El perro será de una determinada raza *(The dog will have a specific breed)*
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario. *(By default it will have "Brown" fur, unless stated otherwise.)*
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario. *(By default it will have no owner, unless stated otherwise.)*
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro. *(It will also have a method called andar (walk), with one input argument (speed_increase). This value will be added to the dog's current speed.)*
* Necesita otro método (parar), donde pondremos la velocidad a 0. *(It needs another method (stop), which sets the speed to 0.)*
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada. *(Another method called "ladrar" (bark), which will have an input argument, and the output will be the string: "GUAU!" + the input argument.)*


Se pide: *(Required:)*
* Implementa la clase *Perro* *(Implement the *Dog* class)*
* Crea un objeto de tipo *Perro*, sin dueño *(Create an object of type *Dog*, without an owner)*
* Comprueba que están bien todos sus atributos *(Check that all its attributes are correct)*
* Prueba que ande, y comprueba su velocidad *(Test that it walks and check its speed)*
* Páralo *(Stop it)*
* Documenta la clase *Perro* *(Document the *Dog* class)*
