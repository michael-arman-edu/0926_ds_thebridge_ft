## Ejercicio 1
*(Exercise 1)*
1. Elimina todos los VW de la siguiente lista. Crea una lista nueva para ello. *(Remove all VW from the following list. Create a new list for it.)*
2. Hazlo de otra manera. Crea una lista nueva, y ve añadiendo todos los que no sean VW *(Do it another way. Create a new list and add all the ones that are not VW)*
3. En vez de `for`, usa un bucle `while` *(Instead of `for`, use a `while` loop)*
> ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 1. Using list comprehension
sin_vw = [car for car in ejer_1 if car != "VW"]
print(sin_vw)
```

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 2. Using for loop and append
sin_vw2 = []
for car in ejer_1:
    if car != "VW":
        sin_vw2.append(car)
print(sin_vw2)
```

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 3. Using while loop
sin_vw3 = []
i = 0
while i < len(ejer_1):
    print(i)
    if ejer_1[i] != "VW":
        sin_vw3.append(ejer_1[i])
    print(sin_vw3)
    i += 1
print("\nFin del bucle:", sin_vw3)
```

## Ejercicio 3
*(Exercise 3)*
1. Escribe un programa que vaya pidiendo numeros al usuario. Cuando el usuario introduzca el 0, el programa tiene que imprimir por pantalla el sumatorio de todos los numeros positivos introducidos *(Write a program that keeps asking the user for numbers. When the user enters 0, the program must print the sum of all positive numbers entered)*
2. Además de la suma, queremos que imprima también una lista con todos los números. *(In addition to the sum, we also want it to print a list with all the numbers.)*

```python
# 1. Sum of positive numbers
total = 0
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    if num > 0:
        total += num
print("El sumatorio de todos los enteros positivos es de", total)
```

```python
# 2. Sum and list of positive numbers
total = 0
numeros = []
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    if num > 0:
        total += num
        numeros.append(num)
print("El sumatorio de todos los enteros positivos fue de", total)
print("Los numeros fueron", numeros)
```

## Ejercicio 4
*(Exercise 4)*
Escribir un programa que vaya pidiendo la cantidad de compras de un cliente. El programa tiene que pedir cantidades hasta que se ingresa un 0. Igualmente cuando una de las cantiaddes introducida es negativa, hay que mandar un mensaje de que "la cantidad es negativa, por favor introducir numeros positivos"

*(Write a program that keeps asking for a customer's purchase amounts. The program must ask for amounts until 0 is entered. Also, when a negative amount is entered, it must display a message saying "the amount is negative, please enter positive numbers")*

```python
total_compras = 0
while True:
    cantidad = float(input("Enter purchase amount (0 to finish): "))
    if cantidad == 0:
        break
    elif cantidad < 0:
        print("Numero negativo, por favor, introduzca un numero positivo: ")
    else:
        total_compras += cantidad
print("Total de compras", total_compras)
```

## Ejercicio 6
*(Exercise 6)*
Escribe un programa que calcule la cantidad de veces que está la letra "m" o "M" en la frase:
*(Write a program that calculates the number of times the letter "m" or "M" appears in the phrase:)*
> "En un lugar de La Mancha, de cuyo nombre no quiero acordarme"

```python
frase = "En un lugar de La Mancha, de cuyo nombre no quiero acordarme"
count = 0
for letra in frase:
    if letra == "m" or letra == "M":
        count += 1
print("Numero de veces que aparece m:", count)
```

## Ejercicio 7
*(Exercise 7)*
Crea un programa que simule la siguiente pirámide. Para ello, el programa tiene que pedirle al usuario un caracter cualquiera. El número de filas es fijo, 5. Vas a necesitar usar dos bucles anidados.

*(Create a program that simulates the following pyramid. For this, the program must ask the user for any character. The number of rows is fixed at 5. You will need to use two nested loops.)*

W W W W W 

W W W W 

W W W 

W W 

W 

```python
caracter = input("Enter a character: ")
filas = 5
for i in range(filas, 0, -1):
    fila = ""
    for j in range(i):
        fila += caracter + " "
    print(fila)
```

## Ejercicio 8
*(Exercise 8)*
Crea un programa que en función de un input numérico que inserte el usuario, imprima por pantalla la pirámide de ejemplo. Si el input es 6, tendrá 6 pisos, pero si introduce 10, la pirámide tendrá 10 pisos.

*(Create a program that, based on a numeric input from the user, prints the example pyramid. If the input is 6, it will have 6 floors, but if 10 is entered, the pyramid will have 10 floors.)*

 1 2 3 4 5 6
 
 1 2 3 4 5
 
 1 2 3 4
 
 1 2 3
 
 1 2
 
 1

```python
n = int(input("Enter the number of floors: "))
for i in range(n, 0, -1):
    fila = ""
    for j in range(1, i + 1):
        fila += " " + str(j)
    print(fila)
```

Imprime toda la secuencia de numeros del 1 al 10, excepto el 3, 4 y 9. Impleméntalo de dos maneras diferentes

*(Print all numbers from 1 to 10, except 3, 4 and 9. Implement it in two different ways)*

```python
# Method 1: using continue
for i in range(1, 11):
    if i in [3, 4, 9]:
        continue
    print(i)

print()

# Method 2: using list comprehension
resultado = [i for i in range(1, 11) if i not in [3, 4, 9]]
print(resultado)
```

## Ejercicio 11
*(Exercise 11)*
Suponiendo que siempre se cumple que las dos listas del ejercicio son iguales de tamaño, implementa un programa que calcule el ratio ingreso/gasto de cada elemento, y lo guarde en una nueva lista. Si hubiese algún problema con los datos, hay que almacenar un mensaje de error en la lista del resultado. Usa `try/except`

*(Assuming both lists are always the same size, implement a program that calculates the income/expense ratio for each element and stores it in a new list. If there is any problem with the data, store an error message in the result list. Use `try/except`)*

> ingresos = [100, 200, 500, 100, 600]
>
> gastos = [50, 20, 70, 0, 25]

```python
ingresos = [100, 200, 500, 100, 600]
gastos = [50, 20, 70, 0, 25]

ratios = []
for i in range(len(ingresos)):
    try:
        ratio = ingresos[i] / gastos[i]
        ratios.append(ratio)
    except ZeroDivisionError as e:
        print(type(e))
        ratios.append(9999)

print(ratios)
```

## Ejercicio 12
*(Exercise 12)*
Muy parecido al anterior, aunque ahora tendremos que excepcionar más tipos de errores con estos nuevos datos. En este caso, almacena un mensaje diferente en la lista resultado, para cada error.

*(Very similar to the previous one, but now we will need to handle more types of errors with this new data. In this case, store a different message in the result list for each error.)*

> ingresos = [100, "HHH", 500, 100, 600, 900, 150]
>
> gastos = [50, 20, 70, 0, 25]

```python
ingresos = [100, "HHH", 500, 100, 600, 900, 150]
gastos = [50, 20, 70, 0, 25]

ratios = []
for i in range(len(ingresos)):
    try:
        ratio = ingresos[i] / gastos[i]
        ratios.append(ratio)
    except TypeError as e:
        print(type(e))
        ratios.append("TypeError")
    except ZeroDivisionError as e:
        print(type(e))
        ratios.append("Zero division error")
    except IndexError as e:
        print(type(e))
        ratios.append("Index Error")

print(ratios)
```
