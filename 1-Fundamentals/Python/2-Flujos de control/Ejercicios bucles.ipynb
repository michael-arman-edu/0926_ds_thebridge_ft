## Ejercicio 1
*(Exercise 1)*
1. Elimina todos los VW de la siguiente lista. Crea una lista nueva para ello. *(Remove all VW from the following list. Create a new list for it.)*
2. Hazlo de otra manera. Crea una lista nueva, y ve añadiendo todos los que no sean VW *(Do it another way. Create a new list and add all the ones that are not VW)*
3. En vez de `for`, usa un bucle `while` *(Instead of `for`, use a `while` loop)*
> ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 1. Usar comprensión de lista para filtrar VW / Use list comprehension to filter out VW
sin_vw = [car for car in ejer_1 if car != "VW"]
print(sin_vw)
```

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 2. Usar bucle for y append para construir la nueva lista / Use for loop and append to build the new list
sin_vw2 = []
for car in ejer_1:
    if car != "VW":
        sin_vw2.append(car)
print(sin_vw2)
```

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]
# 3. Usar bucle while con índice manual / Use while loop with manual index
sin_vw3 = []
i = 0
while i < len(ejer_1):
    print(i)
    if ejer_1[i] != "VW":
        sin_vw3.append(ejer_1[i])
    print(sin_vw3)
    # Incrementar índice para avanzar / Increment index to advance
    i += 1
print("\nFin del bucle:", sin_vw3)
```

## Ejercicio 3
*(Exercise 3)*
1. Escribe un programa que vaya pidiendo numeros al usuario. Cuando el usuario introduzca el 0, imprimir el sumatorio de todos los positivos. *(Keep asking for numbers; when 0 is entered, print the sum of all positive numbers.)*
2. Además imprimir también una lista con todos los números. *(Also print a list with all the numbers.)*

```python
# 1. Acumular la suma de positivos / Accumulate the sum of positive numbers
total = 0
while True:
    num = int(input("Enter a number (0 to stop): "))
    # Salir del bucle si el usuario introduce 0 / Exit the loop if the user enters 0
    if num == 0:
        break
    if num > 0:
        total += num
print("El sumatorio de todos los enteros positivos es de", total)
```

```python
# 2. Acumular suma y lista de positivos / Accumulate sum and list of positive numbers
total = 0
numeros = []
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    if num > 0:
        total += num
        # Añadir el número a la lista / Add the number to the list
        numeros.append(num)
print("El sumatorio de todos los enteros positivos fue de", total)
print("Los numeros fueron", numeros)
```

## Ejercicio 4
*(Exercise 4)*
Pedir cantidades de compra hasta introducir 0; rechazar negativos con mensaje.
*(Ask for purchase amounts until 0 is entered; reject negatives with a message.)*

```python
# Acumular el total de compras / Accumulate the total purchases
total_compras = 0
while True:
    cantidad = float(input("Enter purchase amount (0 to finish): "))
    # Salir si la cantidad es 0 / Exit if amount is 0
    if cantidad == 0:
        break
    elif cantidad < 0:
        # Avisar si la cantidad es negativa / Warn if the amount is negative
        print("Numero negativo, por favor, introduzca un numero positivo: ")
    else:
        total_compras += cantidad
print("Total de compras", total_compras)
```

*(Print the word "Python" to the screen, but starting with the last letter.)*

```python
word = "Python"
# Iterar la cadena invertida con slicing / Iterate the reversed string with slicing
for letra in word[::-1]:
    print(letra)
```

```python
word = "Python"
# Alternativa con while e índice negativo / Alternative with while and negative index
i = len(word) - 1
while i >= 0:
    print(word[i])
    i -= 1
```

## Ejercicio 6
*(Exercise 6)*
Calcular cuántas veces aparece la letra "m" o "M" en la frase.
*(Calculate how many times the letter "m" or "M" appears in the phrase.)*

```python
frase = "En un lugar de La Mancha, de cuyo nombre no quiero acordarme"
count = 0
# Recorrer cada letra y contar m/M / Iterate each letter and count m/M
for letra in frase:
    if letra == "m" or letra == "M":
        count += 1
print("Numero de veces que aparece m:", count)
```

## Ejercicio 7
*(Exercise 7)*
Pirámide descendente con un carácter pedido al usuario, 5 filas fijas.
*(Descending pyramid with a user-supplied character, 5 fixed rows.)*

```python
# Pedir el carácter al usuario / Ask the user for a character
caracter = input("Enter a character: ")
filas = 5
# Bucle externo: filas descendentes / Outer loop: descending rows
for i in range(filas, 0, -1):
    fila = ""
    # Bucle interno: repetir el carácter / Inner loop: repeat the character
    for j in range(i):
        fila += caracter + " "
    print(fila)
```

## Ejercicio 8
*(Exercise 8)*
Pirámide con número de filas dado por el usuario.
*(Pyramid with number of floors given by the user.)*

```python
# Pedir el número de filas al usuario / Ask the user for the number of floors
n = int(input("Enter the number of floors: "))
# Bucle externo: filas decrecientes / Outer loop: decreasing rows
for i in range(n, 0, -1):
    fila = ""
    # Bucle interno: números de 1 hasta i / Inner loop: numbers from 1 to i
    for j in range(1, i + 1):
        fila += " " + str(j)
    print(fila)
```

## Ejercicio 10
*(Exercise 10)*
Imprimir del 1 al 10 excepto el 3, 4 y 9. De dos maneras diferentes.
*(Print 1 to 10 except 3, 4 and 9. In two different ways.)*

```python
# Método 1: usando continue para saltar los excluidos / Method 1: using continue to skip excluded numbers
for i in range(1, 11):
    if i in [3, 4, 9]:
        continue
    print(i)

print()

# Método 2: comprensión de lista filtrando los excluidos / Method 2: list comprehension filtering excluded numbers
resultado = [i for i in range(1, 11) if i not in [3, 4, 9]]
print(resultado)
```

## Ejercicio 11
*(Exercise 11)*
Calcular el ratio ingreso/gasto de cada elemento. Usar `try/except`.
*(Calculate the income/expense ratio for each element. Use `try/except`.)*

```python
ingresos = [100, 200, 500, 100, 600]
gastos = [50, 20, 70, 0, 25]

ratios = []
for i in range(len(ingresos)):
    try:
        # Calcular el ratio ingreso/gasto / Calculate the income/expense ratio
        ratio = ingresos[i] / gastos[i]
        ratios.append(ratio)
    except ZeroDivisionError as e:
        # Manejar división entre cero / Handle division by zero
        print(type(e))
        ratios.append(9999)

print(ratios)
```

## Ejercicio 12
*(Exercise 12)*
Mismo ejercicio pero con más tipos de errores y listas de distinto tamaño.
*(Same exercise but with more error types and lists of different sizes.)*

```python
ingresos = [100, "HHH", 500, 100, 600, 900, 150]
gastos = [50, 20, 70, 0, 25]

ratios = []
for i in range(len(ingresos)):
    try:
        # Calcular el ratio / Calculate the ratio
        ratio = ingresos[i] / gastos[i]
        ratios.append(ratio)
    except TypeError as e:
        # Tipo de dato incorrecto / Wrong data type
        print(type(e))
        ratios.append("TypeError")
    except ZeroDivisionError as e:
        # División entre cero / Division by zero
        print(type(e))
        ratios.append("Zero division error")
    except IndexError as e:
        # Índice fuera de rango (listas de distinto tamaño) / Index out of range (different list sizes)
        print(type(e))
        ratios.append("Index Error")

print(ratios)
```
