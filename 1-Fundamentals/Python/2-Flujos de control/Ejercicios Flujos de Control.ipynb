## Ejercicio 1
*(Exercise 1)*
Declara una variable numérica que será una hora cualquiera del día. Implementa mediante sentencias `if/elif/else` la siguiente casuística:
*(Declare a numeric variable representing any hour of the day. Implement the following cases using `if/elif/else` statements:)*

* Si es entre las 0 y las 8, print "Durmiendo" *(If between 0 and 8, print "Durmiendo")*
* Si es entre las 9 y las 18, print "Trabajando" *(If between 9 and 18, print "Trabajando")*
* Si es entre las 19 y las 21, print "Clase" *(If between 19 and 21, print "Clase")*
* Si es entre las 22 y las 24, print "Descanso" *(If between 22 and 24, print "Descanso")*
* En cualquier otro caso, print "Transporte o error" *(In any other case, print "Transporte o error")*

```python
hora = 23

if 0 <= hora <= 8:
    print("Durmiendo")
elif 9 <= hora <= 18:
    print("Trabajando")
elif 19 <= hora <= 21:
    print("Clase")
elif 22 <= hora <= 24:
    print("Descansando")
else:
    print("Transporte o error")
```

## Ejercicio 2
*(Exercise 2)*
En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo. Tenemos las siguientes variables:
*(In this exercise we will implement a very simple house price calculator. We have the following variables:)*

> superficie
>
> distrito

Implementa mediante sentencias `if/elif/else` la siguiente casuística:
*(Implement the following cases using `if/elif/else` statements:)*

1. Si el distrito es "Moncloa" o "Centro", y además la superficie es superior a 100 metros cuadrados, el precio de la casa es de 1000 *(If the district is "Moncloa" or "Centro", and the area is over 100 square meters, the price is 1000)*
2. Si el distrito es "Salamanca", y además la superficie de la casa es al menos de 150 metros, el precio de la casa es de 1500 *(If the district is "Salamanca", and the area is at least 150 meters, the price is 1500)*
3. Si el distrito no es "Retiro" y la superficie está entre 60 y 80 metros, el precio es de 600 *(If the district is not "Retiro" and the area is between 60 and 80 meters, the price is 600)*
4. En cualquier otro caso, el precio será de 0 *(In any other case, the price will be 0)*

```python
superficie = 120
distrito = "Moncloa"

if (distrito == "Moncloa" or distrito == "Centro") and superficie > 100:
    precio = 1000
elif distrito == "Salamanca" and superficie >= 150:
    precio = 1500
elif distrito != "Retiro" and 60 <= superficie <= 80:
    precio = 600
else:
    precio = 0

print(precio)
```

*(Exercise 3)*

1. Primero se compruebe si el distrito es "Retiro". Si es asi, que imprima "Distrito Retiro", y si no, "Otro distrito" *(First check if the district is "Retiro". If so, print "Distrito Retiro", otherwise "Otro distrito")*
2. Si el distrito es "Retiro", comprueba si si la superficie es mayor de 100 metros cuadrados. En tal caso que imprima un precio de 1000, y si no, de 500. *(If the district is "Retiro", check if the area is greater than 100 square meters. If so, print a price of 1000, otherwise 500.)*

Hay que usar `ifs` anidados *(You must use nested `ifs`)*

```python
superficie = 80
distrito = "Retiro"

if distrito == "Retiro":
    print("Distrito Retiro")
    if superficie > 100:
        print(1000)
    else:
        print(500)
else:
    print("Otro distrito")
```

## Ejercicio 6
*(Exercise 6)*
Dada la siguiente lista:
*(Given the following list:)*

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

*(Implement a program that iterates over the list and prints all numbers divisible by 5. If any number greater than 150 is found, stop the loop.)*

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for num in list1:
    if num > 150:
        break
    if num % 5 == 0:
        print(num)
```

## Ejercicio 8
*(Exercise 8)*
Implementa un programa que imprima por pantalla el siguiente patrón
*(Implement a program that prints the following pattern)*

5 4 3 2 1 

4 3 2 1 

3 2 1 

2 1 

1

NOTA: NO hay lineas en blanco entre una línea y otra. *(NOTE: There are NO blank lines between rows.)*

```python
for i in range(5, 0, -1):
    fila = ""
    for j in range(i, 0, -1):
        fila += str(j) + " "
    print(fila)
```

## Ejercicio 9
*(Exercise 9)*
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:
*(In this exercise we will create a small game. The goal is to guess a number from 1 to 5. We have two attempts. Steps:)*

1. Ya viene implementado cómo obtener un número aleatorio del 1 al 5 *(Getting a random number from 1 to 5 is already implemented)*
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas. *(Declare a lives variable, and use a while loop to check lives remain.)*
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas. *(Inside the loop, get the user's value and check if it's the number. If not, update lives.)*
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose". *(If we guess right, exit the loop and print "You win". If we lose also exit, but print "You lose".)*

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas. *(TIP: it may be useful to use `else` when the `while` loop ends. Whatever is inside will execute once the `while` finishes. You can use it for when you lose.)*

```python
from random import randint
random_number = randint(1,5)
```

```python
vidas = 2
while vidas > 0:
    intento = int(input("Guess a number between 1 and 5: "))
    if intento == random_number:
        print("You win")
        break
    else:
        vidas -= 1
else:
    print("You lose")
```
