## Ejercicio 1
*(Exercise 1)*
Declara una variable numérica que será una hora cualquiera del día. Implementa mediante sentencias `if/elif/else` la siguiente casuística:
*(Declare a numeric variable representing any hour of the day. Implement the following cases using `if/elif/else` statements:)*

* Si es entre las 0 y las 8, print "Durmiendo" *(If between 0 and 8, print "Durmiendo")*
* Si es entre las 9 y las 18, print "Trabajando" *(If between 9 and 18, print "Trabajando")*
* Si es entre las 19 y las 21, print "Clase" *(If between 19 and 21, print "Clase")*
* Si es entre las 22 y las 24, print "Descanso" *(If between 22 and 24, print "Descanso")*
* En cualquier otro caso, print "Transporte o error" *(In any other case, print "Transporte o error")*

```python
# Hora del día a comprobar / Hour of the day to check
hora = 23

# Comprobar en qué franja horaria cae / Check which time slot it falls into
if 0 <= hora <= 8:
    print("Durmiendo")
elif 9 <= hora <= 18:
    print("Trabajando")
elif 19 <= hora <= 21:
    print("Clase")
elif 22 <= hora <= 24:
    print("Descansando")
else:
    print("Transporte o error")
```

## Ejercicio 2
*(Exercise 2)*
En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo.
*(In this exercise we will implement a very simple house price calculator.)*

```python
# Variables de entrada / Input variables
superficie = 120
distrito = "Moncloa"

# Determinar el precio según distrito y superficie / Determine price based on district and area
if (distrito == "Moncloa" or distrito == "Centro") and superficie > 100:
    precio = 1000
elif distrito == "Salamanca" and superficie >= 150:
    precio = 1500
elif distrito != "Retiro" and 60 <= superficie <= 80:
    precio = 600
else:
    precio = 0

print(precio)
```

## Ejercicio 4
*(Exercise 4)*
Escribe un programa que tenga dos variables: un numero, y una lista numérica. El programa debe recorrer la lista e imprimir por pantalla cada elemento de la lista multiplicado por el número

*(Write a program with two variables: a number and a numeric list. The program must iterate over the list and print each element of the list multiplied by the number)*

```python
# Número multiplicador y lista / Multiplier number and list
numero = 2
lista = [1, 2, 3, 4]

# Recorrer la lista y multiplicar cada elemento / Iterate the list and multiply each element
for elem in lista:
    print(elem * numero)
```

## Ejercicio 6
*(Exercise 6)*
Recorre e imprime todos los divisibles por 5. Si alguno es mayor que 150, detener el bucle.
*(Iterate and print all numbers divisible by 5. If any is greater than 150, stop the loop.)*

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for num in list1:
    # Parar si el número es mayor que 150 / Stop if the number is greater than 150
    if num > 150:
        break
    # Imprimir solo los divisibles por 5 / Print only those divisible by 5
    if num % 5 == 0:
        print(num)
```

## Ejercicio 8
*(Exercise 8)*
Implementa un programa que imprima el patrón: 5 4 3 2 1 / 4 3 2 1 / ...
*(Implement a program that prints the pattern: 5 4 3 2 1 / 4 3 2 1 / ...)*

```python
# Bucle externo: controla el número de filas / Outer loop: controls the number of rows
for i in range(5, 0, -1):
    fila = ""
    # Bucle interno: construye cada fila / Inner loop: builds each row
    for j in range(i, 0, -1):
        fila += str(j) + " "
    print(fila)
```

## Ejercicio 9
*(Exercise 9)*
Juego: adivinar un número del 1 al 5 con dos intentos.
*(Game: guess a number from 1 to 5 with two attempts.)*

```python
from random import randint
# Generar número aleatorio entre 1 y 5 / Generate a random number between 1 and 5
random_number = randint(1, 5)
```

```python
# Número de vidas disponibles / Number of lives available
vidas = 2
# Bucle mientras queden vidas / Loop while lives remain
while vidas > 0:
    intento = int(input("Guess a number between 1 and 5: "))
    if intento == random_number:
        print("You win")
        break
    else:
        # Reducir vidas si el intento es incorrecto / Reduce lives if the guess is wrong
        vidas -= 1
else:
    # El else del while se ejecuta si no se usó break / The while's else runs if no break was triggered
    print("You lose")
```
