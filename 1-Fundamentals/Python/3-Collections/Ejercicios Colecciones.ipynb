## Ejercicio 1
*(Exercise 1)*
Dada la siguiente lista: *(Given the following list:)*
> ```ejer_1 = [1,2,3,4,5]```

Inviértela para que quede de la siguiente manera *(Reverse it so it looks like:)*
> ```ejer_1 = [5,4,3,2,1]```

````python
ejer_1 = [1, 2, 3, 4, 5]
ejer_1 = ejer_1[::-1]
print(ejer_1)
````

## Ejercicio 2
*(Exercise 2)*
Eleva todos los elementos de la lista al cuadrado *(Raise all elements of the list to the power of 2)*
> ```ejer_2 = [1,2,3,4,5]```

````python
ejer_2 = [1, 2, 3, 4, 5]
ejer_2 = [x**2 for x in ejer_2]
print(ejer_2)
````

## Ejercicio 3
*(Exercise 3)*
Crea una lista nueva con todas las combinaciones de las siguientes dos listas: *(Create a new list with all combinations of the following two lists:)*
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output: *(Get the following output:)*

['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
combinaciones = [a + " " + b for a in ejer_3_1 for b in ejer_3_2]
print(combinaciones)
````
