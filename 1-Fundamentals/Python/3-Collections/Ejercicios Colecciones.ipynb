## Ejercicio 1
*(Exercise 1)*
Dada la siguiente lista: *(Given the following list:)*
> ```ejer_1 = [1,2,3,4,5]```

Eleva todos los elementos de la lista al cuadrado *(Raise all elements of the list to the power of 2)*

## Ejercicio 3
*(Exercise 3)*
Crea una lista nueva con todas las combinaciones de las siguientes dos listas: *(Create a new list with all combinations of the following two lists:)*
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

## Ejercicio 5
*(Exercise 5)*
Dada la siguiente lista, elimina todos los valores iguales a 3
*(Given the following list, remove all values equal to 3)*
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

Consejo: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.
*(Tip: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.)*

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
# Crear una copia para iterar de forma segura / Create a copy to iterate safely
copia = ejer_5.copy()
# Eliminar cada 3 de la lista original / Remove each 3 from the original list
for elem in copia:
    if elem == 3:
        ejer_5.remove(elem)
print(ejer_5)
````

## Ejercicio 6
*(Exercise 6)*
1. Crea una tupla con 3 elementos *(Create a tuple with 3 elements)*
2. Crea otra tupla con un elemento y comprueba su tipo *(Create another tuple with one element and check its type)*
3. Crea una tupla con elementos de diferentes tipos *(Create a tuple with elements of different types)*
4. Imprime por pantalla el primer y ultimo elemento de la tupla del apartado 3. Usa `len` para el ultimo *(Print the first and last element of the tuple from point 3. Use `len` for the last one)*

## Ejercicio 12
*(Exercise 12)*
Convierte la siguiente tupla en un diccionario
*(Convert the following tuple into a dictionary)*
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.
*(TIP: You can use two control variables in a `for` loop. One for each value of the tuples.)*

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
dict_12 = {}
# Iterar con dos variables de control / Iterate with two control variables
# Las claves repetidas sobreescriben el valor anterior / Repeated keys overwrite the previous value
for clave, valor in ejer_12:
    dict_12[clave] = valor
print(dict_12)
````

````python
# Alternativa con comprensión de diccionario / Alternative with dictionary comprehension
dict_12 = {clave: valor for clave, valor in ejer_12}
print(dict_12)
````

````python
# Acceder a un valor por su clave / Access a value by its key
print(dict_12["x"])
````

## Ejercicio 13
*(Exercise 13)*
1. Crea una lista ordenada ascendente con las claves del diccionario *(Create a sorted ascending list with the dictionary keys)*
2. Crea otra lista ordenada descendente con los valores *(Create another sorted descending list with the values)*
3. Añade una nueva clave/valor *(Add a new key/value)*
4. Busca la clave 2 dentro del diccionario *(Look for key 2 in the dictionary)*
5. Itera la clave y el valor del diccionario con un unico for *(Iterate the key and value of the dictionary with a single for)*
> ```ejer_13 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_13 = {4: 78, 2: 98, 8: 234, 5: 29}

# 1. Claves ordenadas de forma ascendente / Keys sorted in ascending order
print(sorted(ejer_13.keys()))

# 2. Valores ordenados de forma descendente / Values sorted in descending order
print(sorted(ejer_13.values(), reverse=True))

# 3. Añadir una nueva clave/valor / Add a new key/value pair
ejer_13[9] = 65
print(ejer_13)

# 4. Comprobar si la clave 2 existe / Check if key 2 exists
print(2 in ejer_13)

# 5. Iterar clave y valor con .items() / Iterate key and value with .items()
for clave, valor in ejer_13.items():
    print(f"Clave: {clave}, Valor -> {valor}")
````

## Ejercicio 16
*(Exercise 16)*
Multiplica todos los valores del diccionario
*(Multiply all values of the dictionary)*
> ```ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}
# Acumular el producto iterando los valores / Accumulate the product by iterating the values
resultado = 1
for v in ejer_16.values():
    resultado *= v
print(resultado)
````
